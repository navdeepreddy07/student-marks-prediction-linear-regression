```python
import numpy as np

# Input feature: Study hours
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)

# Target: Student marks
y = np.array([35, 42, 50, 58, 67])

print(X)
```

```python

import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python

# 2. Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=60
)
```

```python
# 1. Prepare the data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([35, 42, 50, 58, 67])
```

```python

# 3. Create and train the model
model = LinearRegression()
model.fit(X_train, y_train)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python

# 4. Evaluate on test data
y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print("Actual marks:", y_test)
print("Predicted marks:", y_pred)
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
```

```python
# 5. Predict for a new student
new_student = np.array([[6]])
prediction = model.predict(new_student)
print("Predicted marks for 6 hours:", prediction[0])
```

```python
# 6. Visualize actual data and regression line
plt.scatter(X, y, label="Actual Marks")
plt.plot(X, model.predict(X), label="Prediction Line")

plt.xlabel("Study Hours")
plt.ylabel("Student Marks")
plt.title("Student Marks Prediction - Linear Regression")
plt.legend()
plt.grid(True)
plt.show()
```
